```python
# =============================================
# PASO 1: CONFIGURACIÓN DEL ENTORNO Y CARGA
# =============================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Cargar el dataset
df = pd.read_csv('Proyecto_SmartFactory_Falla_Equipos.csv')
print("Dataset cargado correctamente ✓")
```

```python
# =============================================
# PASO 2: ENTENDIMIENTO DE LOS DATOS
# =============================================

# Dimensiones del dataset
print(f"Filas: {df.shape[0]} | Columnas: {df.shape[1]}")

# Primeras 10 filas
df.head(10)
```

```python
# Tipos de datos por columna
print("=== TIPOS DE DATOS ===")
print(df.dtypes)

# Valores nulos por columna
print("\n=== VALORES NULOS ===")
print(df.isnull().sum())
```

```python
# =============================================
# INSIGHT DE NEGOCIO: Estrategia de imputación
# =============================================

"""
Si existieran nulos en los sensores de temperatura (Air/Process temperature),
se recomendaría imputar con la MEDIANA en lugar de la media.
Razón: Los sensores industriales pueden registrar picos o lecturas erróneas
(outliers) que distorsionan la media. La mediana es más robusta frente a
valores atípicos y representa mejor el comportamiento típico del proceso.
"""
print("Insight documentado ✓")
```

Comentario: Calidad Estructural

- El dataset **no presenta valores nulos** en ninguna columna.
- Los tipos de datos son correctos:
  - `float64`: sensores numéricos (temperaturas, velocidad, torque, desgaste)
  - `object`: columnas de texto (Type, Product ID)
  - `int64`: variables binarias (Machine failure, TWF, HDF, PWF, OSF, RNF)
- No se requiere corrección de tipos ni imputación en este dataset.

Insight de Negocio: Estrategia de Imputación en Sensores de Temperatura

Si existieran valores nulos en `Air temperature [K]` o `Process temperature [K]`,
se recomendaría imputar con la **mediana** en lugar de la media.

**Razón:** Los sensores industriales pueden registrar picos o lecturas erróneas
que generan outliers. La mediana es robusta frente a valores extremos y representa
mejor el comportamiento típico del proceso, evitando que un valor anómalo
distorsione la imputación.

```python
# =============================================
# PASO 3: ANÁLISIS DE NEGOCIO
# =============================================

plt.figure(figsize=(8, 5))
plt.hist(df['Rotational speed [rpm]'], bins=30, color='steelblue', edgecolor='black')
plt.title('Distribución de Velocidad de Rotación [rpm]')
plt.xlabel('RPM')
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.show()

# Estadísticas descriptivas del campo
print(df['Rotational speed [rpm]'].describe().round(2))
```

Nota Técnica: Intervalo de Mayor Frecuencia - Rotational Speed

Del histograma se observa que la mayor concentración de registros se ubica
aproximadamente entre **1300 y 1600 rpm**.

Esto representa el **rango de operación normal** del equipo: la mayoría de las
máquinas trabajan establemente en ese intervalo. Valores fuera de este rango
(muy bajos o muy altos) pueden ser indicadores de condiciones anómalas,
sobrecarga o falla inminente del equipo.

```python
# Frecuencia absoluta y porcentual
freq_abs = df['Machine failure'].value_counts().sort_index()
freq_pct = df['Machine failure'].value_counts(normalize=True).sort_index() * 100

print("=== FRECUENCIA ABSOLUTA ===")
print(freq_abs)
print("\n=== FRECUENCIA PORCENTUAL ===")
print(freq_pct.round(2))

# Gráfico de barras (frecuencia absoluta) y torta (porcentual)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Barras
axes[0].bar(['Normal (0)', 'Falla (1)'], freq_abs.values,
            color=['steelblue', 'tomato'], edgecolor='black')
axes[0].set_title('Frecuencia Absoluta - Machine Failure')
axes[0].set_ylabel('Cantidad de registros')
axes[0].set_xlabel('Estado del equipo')
for i, v in enumerate(freq_abs.values):
    axes[0].text(i, v + 30, str(v), ha='center', fontweight='bold')

# Torta
axes[1].pie(freq_pct.values,
            labels=['Normal (0)', 'Falla (1)'],
            autopct='%1.2f%%',
            colors=['steelblue', 'tomato'],
            startangle=90,
            explode=(0, 0.05))
axes[1].set_title('Proporción - Machine Failure')

plt.tight_layout()
plt.show()
```

Insight de Negocio: El Desbalanceo como Desafío para la Minería de Datos

El dataset presenta un fuerte desbalanceo: aproximadamente **96-97% de registros
normales** frente a solo un **3-4% de fallas**.

Esto es un desafío porque:

1. **El modelo aprende a ignorar la clase minoritaria:** Un clasificador que prediga
   siempre "Normal" obtendría ~97% de exactitud sin detectar ninguna falla real.
2. **La métrica de exactitud (accuracy) es engañosa:** Un modelo puede parecer muy
   preciso pero ser completamente inútil para el negocio.
3. **El costo real es asimétrico:** No detectar una falla (falso negativo) tiene un
   costo operacional mucho mayor que una falsa alarma.

**Técnicas para compensarlo:** SMOTE (sobremuestreo sintético), `class_weight='balanced'`
en el modelo, o ajuste del umbral de decisión.

```python
# =============================================
# PASO 4: INGENIERÍA DE CARACTERÍSTICAS
# =============================================

df['Temp_diff'] = df['Process temperature [K]'] - df['Air temperature [K]']
print("Columna Temp_diff creada ✓")

# Verificación
df[['Air temperature [K]', 'Process temperature [K]', 'Temp_diff']].head(10)
```

Nota Técnica: Nuevo Conocimiento de Temp_diff

`Temp_diff` captura el **gradiente térmico** del proceso industrial.

Las temperaturas por separado solo indican valores absolutos, pero no informan
cuánto está **calentando** la máquina en relación al ambiente. Por ejemplo:

- Una `Process temperature` de 310K parece normal...
- ...pero si el `Air temperature` es 309K, el gradiente es mínimo (máquina fría)
- Si el `Air temperature` es 295K, el gradiente es 15K (máquina muy caliente)

Un `Temp_diff` elevado puede anticipar **sobrecalentamiento inminente**, incluso
cuando ambas temperaturas absolutas lucen normales por separado. Esta columna
aporta información relacional que ninguna variable individual entregaba.

```python
# =============================================
# PASO 5: TRATAMIENTO DE OUTLIERS
# =============================================

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(y=df['Rotational speed [rpm]'], ax=axes[0], color='lightblue')
axes[0].set_title('Boxplot - Rotational Speed [rpm]')
axes[0].set_ylabel('RPM')

sns.boxplot(y=df['Torque [Nm]'], ax=axes[1], color='lightcoral')
axes[1].set_title('Boxplot - Torque [Nm]')
axes[1].set_ylabel('Nm')

plt.suptitle('Detección Visual de Outliers', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

Nota Técnica: Outliers y Elección de Técnica de Escalado

Ambos boxplots muestran **valores atípicos** (puntos fuera de los bigotes):

- `Rotational speed [rpm]`: outliers hacia valores altos (sobre ~2500 rpm)
- `Torque [Nm]`: outliers en ambos extremos

**Impacto en el escalado:**
- `MinMaxScaler` es muy sensible a outliers: un solo valor extremo comprime
  todo el rango útil de los datos hacia cero, distorsionando la escala.
- `StandardScaler` usa media y desviación estándar, siendo más robusto ante
  valores extremos.

**Decisión:** Se aplicará `StandardScaler` a `Rotational speed` y `Torque`,
y `MinMaxScaler` a las variables sin outliers severos.

```python
# =============================================
# PASO 6: PREPARACIÓN DE DATOS
# =============================================

# One Hot Encoding a la columna Type
df = pd.get_dummies(df, columns=['Type'], drop_first=False)

# Convertir columnas bool a int (compatibilidad con sklearn)
bool_cols = [col for col in df.columns if df[col].dtype == bool]
df[bool_cols] = df[bool_cols].astype(int)

print("One Hot Encoding aplicado ✓")
print("Columnas generadas:", [c for c in df.columns if 'Type_' in c])

# Prevención de Data Leakage
cols_leakage = ['TWF', 'HDF', 'PWF', 'OSF', 'RNF']
cols_id      = ['UDI', 'Product ID']

df.drop(columns=cols_leakage + cols_id, inplace=True)

print("\nColumnas eliminadas:", cols_leakage + cols_id)
print("Columnas restantes:", list(df.columns))
```

Defensa de Integridad: ¿Por qué eliminar TWF, HDF, PWF, OSF, RNF?

Estas columnas indican el **tipo específico de falla** que ocurrió
(desgaste de herramienta, disipación de calor, potencia, etc.).

**El problema:** Esta información solo existe **después de que la falla ya ocurrió**.
En un escenario real de predicción, cuando queremos anticipar si un equipo fallará,
estas columnas no estarían disponibles, ya que la falla aún no ha sucedido.

Si el modelo se entrena con ellas:
- Aprendería que "si TWF=1 entonces Machine failure=1", lo que es trivial y circular.
- Obtendría una precisión artificialmente perfecta (~100%) en entrenamiento.
- En producción real, sin esas columnas, el modelo colapsaría completamente.

**Conclusión:** Sí, mantenerlas **invalidaría totalmente el propósito predictivo**
del modelo. El objetivo es predecir fallas ANTES de que ocurran, no describirlas
una vez que ya pasaron.

```python
# =============================================
# PASO 7: ESTRUCTURACIÓN FINAL
# =============================================

y = df['Machine failure']
X = df.drop(columns=['Machine failure'])

print("Vector objetivo  y:", y.shape)
print("Matriz de características X:", X.shape)
print("\nColumnas de X:")
print(list(X.columns))
```

```python
# =============================================
# PASO 8: ESCALAMIENTO MIXTO
# =============================================

# --- StandardScaler: velocidad y torque (tienen outliers) ---
scaler_std = StandardScaler()
cols_std = ['Rotational speed [rpm]', 'Torque [Nm]']
X = X.copy()
X[cols_std] = scaler_std.fit_transform(X[cols_std])

# --- MinMaxScaler: columnas restantes sin outliers severos ---
scaler_mm = MinMaxScaler()
cols_mm = ['Air temperature [K]', 'Process temperature [K]', 'Tool wear [min]', 'Temp_diff']
X[cols_mm] = scaler_mm.fit_transform(X[cols_mm])

print("Escalamiento aplicado ✓")
```

Justificación Técnica: ¿Por qué StandardScaler en velocidad y torque?

`Rotational speed` y `Torque` presentaron **outliers significativos** en los boxplots.
`StandardScaler` transforma los datos para que tengan **media=0 y desviación estándar=1**,
siendo robusto frente a valores extremos.

`MinMaxScaler` llevaría todos los valores al rango [0,1], pero un solo outlier
extremo comprimiría el resto de los datos hacia un rango muy pequeño, perdiendo
la variabilidad útil.

**¿Por qué no StandardScaler en Tool wear?**
`Tool wear [min]` es un contador acumulado de tiempo (0 a ~250 min) sin dispersión
extrema ni outliers severos. `MinMaxScaler` lo normaliza limpiamente al rango [0,1],
preservando su distribución original de forma óptima.

```python
# =============================================
# VERIFICACIÓN FINAL
# =============================================

print("=" * 55)
print("StandardScaler — media ≈ 0, std ≈ 1")
print("=" * 55)
print(X[cols_std].describe().round(3))

print("\n" + "=" * 55)
print("MinMaxScaler — min = 0, max = 1")
print("=" * 55)
print(X[cols_mm].describe().round(3))

print("\n" + "=" * 55)
print("Columnas One Hot Encoding (rango 0-1)")
print("=" * 55)
type_cols = [c for c in X.columns if 'Type_' in c]
print(X[type_cols].describe().round(3))

print("\n✅ Dataset listo para modelado")
print(f"   X shape: {X.shape} | y shape: {y.shape}")
```
